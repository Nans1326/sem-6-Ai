Task1 - Unit Converter Program
Step 1: Create Conversion Function


```python
def convert_units(conversion_type, value):
    """
    This function converts units of length, weight, and volume.

    Parameters:
    conversion_type (str): type of conversion
    value (float): value to convert

    Returns:
    float: converted value
    """

    if conversion_type == "m_to_ft":
        return value * 3.28084

    elif conversion_type == "ft_to_m":
        return value / 3.28084

    elif conversion_type == "kg_to_lbs":
        return value * 2.20462

    elif conversion_type == "lbs_to_kg":
        return value / 2.20462

    elif conversion_type == "l_to_gal":
        return value * 0.264172

    elif conversion_type == "gal_to_l":
        return value / 0.264172

    else:
        return "Invalid conversion type"
```

Step 2: Take User Input

```python
try:
    print("Choose Conversion Type")
    print("1: meters to feet")
    print("2: feet to meters")
    print("3: kg to pounds")
    print("4: pounds to kg")
    print("5: liters to gallons")
    print("6: gallons to liters")

    choice = int(input("Enter choice number: "))
    value = float(input("Enter value: "))

    if choice == 1:
        result = convert_units("m_to_ft", value)

    elif choice == 2:
        result = convert_units("ft_to_m", value)

    elif choice == 3:
        result = convert_units("kg_to_lbs", value)

    elif choice == 4:
        result = convert_units("lbs_to_kg", value)

    elif choice == 5:
        result = convert_units("l_to_gal", value)

    elif choice == 6:
        result = convert_units("gal_to_l", value)

    else:
        result = "Invalid choice"

    print("Converted value:", result)

except:
    print("Error: Please enter valid numbers")
```

TASK 2 - MAthematical Operations on List
Step 1: create functions

```python
def find_sum(numbers):
    """Returns sum of numbers"""
    return sum(numbers)

def find_average(numbers):
    """Returns average"""
    return sum(numbers) / len(numbers)

def find_max(numbers):
    """Returns maximum number"""
    return max(numbers)

def find_min(numbers):
    """Returns minimum number"""
    return min(numbers)
```

Step 2: User Input program

```python
try:
    nums = input("Enter numbers separated by space: ")
    numbers = list(map(float, nums.split()))

    print("Choose Operation")
    print("1 Sum")
    print("2 Average")
    print("3 Maximum")
    print("4 Minimum")

    choice = int(input("Enter choice: "))

    if choice == 1:
        print("Sum =", find_sum(numbers))

    elif choice == 2:
        print("Average =", find_average(numbers))

    elif choice == 3:
        print("Maximum =", find_max(numbers))

    elif choice == 4:
        print("Minimum =", find_min(numbers))

    else:
        print("Invalid choice")

except:
    print("Invalid input")
```

7. Flatten a list of lists

```python
# Function to flatten a deeply nested list
def deep_flatten(lst):

    result = []          # Final flat list
    stack = list(lst)    # Copy input into a stack

    # Process until stack is empty
    while stack:

        item = stack.pop(0)   # Take first element

        if type(item) == list:
            # If it is a list → open it and add items back to stack
            stack = item + stack
        else:
            # If it is a number → add to result
            result.append(item)

    return result


# Example input
data = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]

# Call function
print(deep_flatten(data))
```

8. Nested list Average

```python
# Function to find average of nested list
def average_nested(lst):

    total = 0      # to store sum of all numbers
    count = 0      # to count how many numbers

    # Go through each sublist
    for sublist in lst:

        # Go through each number inside sublist
        for num in sublist:
            total += num   # add number to total
            count += 1     # increase count

    # Calculate average
    return total / count


# Example input
data = [[1,2],[3,4],[5,6]]

# Print result
print(average_nested(data))
```

Problem 2: Array Manipulation, Ranges, indexing

```python
# Import NumPy
import numpy as np


# 1️⃣ Create array from 1 to 10
arr = np.arange(1, 11)
print("Original array:", arr)


# 2️⃣ Get first 5 elements
first_five = arr[:5]
print("First 5 elements:", first_five)


# 3️⃣ Get last 3 elements
last_three = arr[-3:]
print("Last 3 elements:", last_three)


# 4️⃣ Get elements from index 2 to 6
middle_part = arr[2:7]
print("Index 2 to 6:", middle_part)


# 5️⃣ Change value at index 0
arr[0] = 100
print("After changing first element:", arr)


# 6️⃣ Select every second element
every_second = arr[::2]
print("Every second element:", every_second)
```

```python
# Import NumPy
import numpy as np


# Create two arrays
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])


# 1️⃣ Addition
add = a + b
print("Addition:", add)


# 2️⃣ Subtraction
sub = b - a
print("Subtraction:", sub)


# 3️⃣ Multiplication
mul = a * b
print("Multiplication:", mul)


# 4️⃣ Division
div = b / a
print("Division:", div)


# 5️⃣ Square of array a
square = a ** 2
print("Square of a:", square)


# 6️⃣ Sum of elements
print("Sum of a:", np.sum(a))


# 7️⃣ Maximum value
print("Max of b:", np.max(b))


# 8️⃣ Minimum value
print("Min of b:", np.min(b))
```

Problem 4: matrix Operations

```python
# Import NumPy
import numpy as np


# Create two matrices (2D arrays)
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])


# 1️⃣ Matrix Addition
add = A + B
print("Addition:\n", add)


# 2️⃣ Matrix Subtraction
sub = A - B
print("Subtraction:\n", sub)


# 3️⃣ Element-wise Multiplication
mul = A * B
print("Element-wise Multiplication:\n", mul)


# 4️⃣ Matrix Multiplication (Dot Product)
mat_mul = np.dot(A, B)
print("Matrix Multiplication:\n", mat_mul)


# 5️⃣ Transpose of matrix A
transpose = A.T
print("Transpose of A:\n", transpose)


# 6️⃣ Determinant of matrix A
det = np.linalg.det(A)
print("Determinant of A:", det)
```
